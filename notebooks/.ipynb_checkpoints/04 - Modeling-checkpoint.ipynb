```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import FunctionTransformer, StandardScaler, MinMaxScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer

from sklearn.linear_model import LinearRegression

from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, root_mean_squared_error, r2_score
```

```python
models_performance_dict, y_pred_dict = dict(), dict()
```

# First approach: year of 2016, residential (type 1)

```python
df_2016_type1 = df[
    (df.ano == 2016) & (df.ID_UTILIZACAO == 1)
]
df_2016_type1.describe().T
```

```python
df_2016_type1.isna().sum()
```

```python
print("Number of:")
print("  Typology:", df_2016_type1.COD_TIPOLOGIA.nunique())
print("  Neighborhoods:", df_2016_type1.COD_BAIRRO.nunique())
print("  Streets:", df_2016_type1.COD_LOGRADOURO.nunique())
print("  Passage:", df_2016_type1.COD_TRECHO.nunique())
```

```python
def calc_scores(model_name, y_pred_train, y_pred_test, X_train, X_test, y_train, y_test, add_to_dict=True):
    
    # store predictions
    y_pred_dict[(model_name, "train")] = y_pred_train.copy()
    y_pred_dict[(model_name, "test")] = y_pred_test.copy()

    # train scores
    MAE_train = mean_absolute_error(y_train, y_pred_train, multioutput="raw_values")
    MSE_train = mean_squared_error(y_train, y_pred_train)
    RMSE_train = root_mean_squared_error(y_train, y_pred_train)
    R2_train = r2_score(y_train, y_pred_train)
    
    # test scores
    MAE_test = mean_absolute_error(y_test, y_pred_test, multioutput="raw_values")
    MSE_test = mean_squared_error(y_test, y_pred_test)
    RMSE_test = root_mean_squared_error(y_test, y_pred_test)
    R2_test = r2_score(y_test, y_pred_test) 

    if add_to_dict:

        # train scores
        models_performance_dict[(model_name, "MAE" , "train")] = MAE_train
        models_performance_dict[(model_name, "MSE" , "train")] = MSE_train
        models_performance_dict[(model_name, "RMSE", "train")] = RMSE_train
        models_performance_dict[(model_name, "R2"  , "train")] = R2_train
        
        # test scores
        models_performance_dict[(model_name, "MAE" , "test")] = MAE_test
        models_performance_dict[(model_name, "MSE" , "test")] = MSE_test
        models_performance_dict[(model_name, "RMSE", "test")] = RMSE_test
        models_performance_dict[(model_name, "R2"  , "test")] = R2_test

    else:
        return MAE_train, MSE_train, RMSE_train, R2_train, MAE_test, MSE_test, RMSE_test, R2_test
```

## Linear regression

```python
X, y = df_2016_type1.drop(["valor_m2", "vl_declarado"], axis=1), df_2016_type1.valor_m2
X_train, X_test, y_train, y_test = train_test_split(X, y)
```

```python
preprocessor = ColumnTransformer([
    (
        "location",
        OneHotEncoder(handle_unknown="ignore"),
        ["COD_TIPOLOGIA", "COD_BAIRRO","COD_LOGRADOURO"]
    ),
    (
        "log_area",
        Pipeline([
            ("log", FunctionTransformer(np.log, inverse_func=np.exp, feature_names_out="one-to-one")), # convert to log space
            ("scaler", StandardScaler())
        ]),
        ["area"]
    ),
    (
        "age_factor",
        Pipeline([
            ("scaler", MinMaxScaler())
        ]),
        ["FATOR_IDADE"]
    )
])
```

```python
lr1_model = Pipeline([
    ("preprocesor", preprocessor),
    ("regressor", LinearRegression())
])
lr1_model.fit(X_train, y_train)
```

```python
calc_scores(
    "LR-1",
    lr1_model.predict(X_train),
    lr1_model.predict(X_test),
    X_train, X_test, y_train, y_test
)

pd.DataFrame.from_dict(models_performance_dict, orient="columns").T.unstack([2, 1])
```

Let'see if prediting the totla declared value, instead of value per square meter, the model can reach better performance.

```python
lr2_model = Pipeline([
    ("preprocesor", preprocessor),
    ("regressor", LinearRegression())
])
lr2_model.fit(X_train, np.log(y_train))
```

```python
calc_scores(
    "LR-2",
    np.exp(lr2_model.predict(X_train)),
    np.exp(lr2_model.predict(X_test)),
    X_train, X_test, y_train, y_test
)

pd.DataFrame.from_dict(models_performance_dict, orient="columns").T.unstack([2, 1])
```

Well... slightly better results, but not significant. 
Let's see if using the total declared value, represented by `vl_declarado`, we can get better results.
This might impact positively the model's performance because the area is a value prone to mistakes.

```python
X, y = df_2016_type1.drop(["valor_m2", "vl_declarado"], axis=1), df_2016_type1.vl_declarado
X_train, X_test, y_train, y_test = train_test_split(X, y)
```

```python
lr3_model = Pipeline([
    ("preprocesor", preprocessor),
    ("regressor", LinearRegression())
])
lr3_model.fit(X_train, y_train)
```

```python
calc_scores(
    "LR-3",
    lr3_model.predict(X_train),
    lr3_model.predict(X_test),
    X_train, X_test, y_train, y_test
)

pd.DataFrame.from_dict(models_performance_dict, orient="columns").T.unstack([2, 1])
```

Not much difference. Let's see the using log-space

```python
lr4_model = Pipeline([
    ("preprocesor", preprocessor),
    ("regressor", LinearRegression())
])
lr4_model.fit(X_train, np.log(y_train))
```

```python
calc_scores(
    "LR-4",
    np.exp(lr4_model.predict(X_train)),
    np.exp(lr4_model.predict(X_test)),
    X_train, X_test, y_train, y_test
)
pd.DataFrame.from_dict(models_performance_dict, orient="columns").T.unstack([2, 1])
```

Now we have something.
Train and test R² scores are ~0.83 and ~0.75, respectively.
This is an indication of overfitting, but it is not worrying for this first approach with linear regression.

```python
fig, ax = plt.subplots()
ax.scatter(y_test, y_pred_dict[("LR-4", "test")], marker=".")
ax.set_xscale("log")
ax.set_yscale("log")

ax.set_title("LR-4: linear regression")
ax.set_xlabel("Declared value (R$)")
ax.set_ylabel("Predicted value (R$)")
ax.grid()
```

Now let's check quickly the influence of each parameter on the prediction of the property value.

```python
from sklearn.inspection import permutation_importance

result = permutation_importance(
    lr4_model,
    X,
    np.log(y),
    scoring="explained_variance",
)

df_importance = pd.DataFrame({
    "feature": X.columns,
    "importance": result.importances_mean,
    "std": result.importances_std
}).sort_values("importance", ascending=False).set_index("feature")

df_importance
```

By the table above, the neighborhood (`COD_BAIRRO`) is clearly the most influential feature on the property value.
More than the area itself.

The street (`COD_LOGRADOURO`) has a similar influence, followed by the age (`FATOR_IDADE`) of the property.

Comparing the models with the 3 best scores:
- **Ridge Regression**: Fast to train and shows a small train–test gap, indicating good stability. However, its lower overall scores suggest that a linear model cannot fully capture the nonlinear relationships and interactions present in the data.
- **Random Forest**: Achieved the best predictive performance, with the highest test (R2) and lowest MAE/RMSE. However, it presents a larger train–test gap and a considerably higher training time, suggesting higher variance and greater computational cost.
- **XGBoost**: Achieved performance close to Random Forest while training much faster and maintaining a smaller train–test gap. It can model nonlinear relationships and feature interactions while offering explicit regularization and efficient hyperparameter tuning.

**XGBoost was therefore selected as the main model for further development**, as it provides the best compromise between predictive performance, generalization, computational efficiency, and scalability for the larger and more heterogeneous dataset that will be used in later stages.

## Cross-validation

Cross-validation will be used to obtain a more reliable estimate of the model's performance and reduce the influence of randomness in the train-test split.


```python
from sklearn.model_selection import RepeatedKFold, cross_validate
```

```python
# total property value of 2016
X, y = df_2016_type1.drop(["valor_m2", "vl_declarado"], axis=1), df_2016_type1.vl_declarado
```

```python
cv = RepeatedKFold(
    n_splits=5,
    n_repeats=10,
    random_state=121289
)

_scores = cross_validate(
    ml_model,
    X,
    y,
    cv=cv,
    scoring={
        "mae": "neg_mean_absolute_error",
        "rmse": "neg_root_mean_squared_error",
        "r2": "r2"
    },
    n_jobs=-1,
    return_train_score=True,
    return_indices=False,
    verbose=2,
)
```

```python
pd.DataFrame(_scores).describe()
```

Cross-validation resulted in an average validation R2 of 0.903 with a standard deviation of 0.039, while the training score remained highly stable at (0.964 +- 0.002). 
Low variability in validation performance indicates that the model is **not** sensitive to the composition of the validation subset.

## Consider all time series

Now let's consider the entire dataset for all years.
To avoid data leakage and account data drift, the model is trained differently:
- The dataset of a given year is trained with the data of all previous years. Thus, for a given year $t$, train data is for previous year $t-1$ and test data is for $t$.
- The value is normalized with the mean value of the previous years.

```python
df["date"] = pd.to_datetime(
    dict(year=df["ano"], month=df["mes"], day=1)
)
```

```python
results_movingtime_dict = dict()
for pred_date in pd.date_range(
    start="2017-01-01",
    end="2026-03-01",
    freq="MS"   # Month Start
):
    # time slices
    train_start = pred_date - pd.DateOffset(months=24)
    test_end = pred_date + pd.DateOffset(months=1)
    
    # train and test data is relative to time
    df_train = df[(df.date >= train_start) & (df.date < pred_date) & (df.ID_UTILIZACAO == 1)]
    df_test = df[(df.date >= pred_date) & (df.date <= test_end) & (df.ID_UTILIZACAO == 1)]

    # train-test split with the total value
    X_train, y_train = df_train.drop(["valor_m2", "vl_declarado"], axis=1), df_train.vl_declarado
    X_test, y_test   = df_test.drop(["valor_m2", "vl_declarado"], axis=1),  df_test.vl_declarado

    # normalization
    # in log space: log(value/ref) = log(value) - log(ref)
    norm_train_value = y_train.median()
    y_train /= norm_train_value
    y_test /= norm_train_value
    
    # trainning
    ml_model.fit(X_train, np.log(y_train))

    # predition
    pred_train = ml_model.predict(X_train)
    pred_test = ml_model.predict(X_test)
    
    # calculate scores
    _temp = calc_scores(
        f"XGBoost 4 : {pred_date.strftime("%Y-%m")}",
        np.exp(ml_model.predict(X_train)),
        np.exp(ml_model.predict(X_test)),
        X_train, X_test, y_train, y_test,
        add_to_dict=False
    )
    results_movingtime_dict[pred_date.strftime("%Y-%m")] = _temp
```

```python
df_results = pd.DataFrame.from_dict(
    results_movingtime_dict, 
    orient="index", 
    columns=["train_MAE", "train_MSE", "train_RMSE", "train_R2", "test_MAE", "test_MSE", "test_RMSE", "test_R2"]
)
df_results["R2_gap"] = df_results.train_R2 - df_results.test_R2
df_results.describe().T
```

```python
import matplotlib.dates as mdates

fig, ax = plt.subplots(figsize=(12,7))

df_results.plot(y=["train_R2", "test_R2"], grid=True, ax=ax)

#ax.xaxis.set_major_locator(mdates.YearLocator())
#ax.xaxis.set_major_formatter(mdates.DateFormatter("%Y"))
```
